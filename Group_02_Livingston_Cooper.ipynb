```python
#Question: Which time of the year sees the most money put into games?
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Group by 'MONTH' and calculate the average 'COIN_IN'
average_coins_per_month = df.groupby('MONTH')['COIN_IN'].mean().reset_index()
average_coins_per_month.rename(columns={'COIN_IN': 'AVG_COINS_PER_MONTH'}, inplace=True)

# Create the bar chart
plt.figure(figsize=(10, 5))
sns.barplot(x='MONTH', y='AVG_COINS_PER_MONTH', data=average_coins_per_month, color='grey')

# Customize the chart
plt.xlabel("Month")
plt.ylabel("Average Coins")
plt.title("Average Coins Per Month")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()

```

```python
#Question: What are the Top 10 Games (GAME_TITLE) based on total handle pulls (HANDLE_PULLS) ?
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.ticker as ticker

# Group by GAME_TITLE and sum HANDLE_PULLS
top_games = df.groupby('GAME_TITLE')['HANDLE_PULLS'].sum().reset_index()

# Sort in descending order and take the top 10
top_games = top_games.sort_values(by='HANDLE_PULLS', ascending=False).head(10)

# Create the bar chart
plt.figure(figsize=(12, 6))
ax = sns.barplot(x='GAME_TITLE', y='HANDLE_PULLS', data=top_games, color='grey')

# Format y-axis in millions
ax.yaxis.set_major_formatter(ticker.FuncFormatter(lambda x, _: f'{x / 1e6:.1f}M'))

# Customize the chart
plt.xlabel("Game Title")
plt.ylabel("Total Handle Pulls")
plt.title("Top 10 Game Titles by Total Handle Pulls")
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the plot
plt.show()




```
